```python
# Importing Major Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# additional library

import plotly.express as px

import warnings
warnings.filterwarnings('ignore')
```

```python
# importing datasets

sns.get_dataset_names()

tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")
mpg = sns.load_dataset('mpg')

gap = px.data.gapminder()
tips
titanic
mpg
gap
```

```python
# scatterplot
# bivariate analysis
# numerical vs numerical plot
# clusters
# relationship
# anomalies
# outliers detect
# patterns

iris
# size manipulate

plt.figure(figsize=(8,3))
sns.scatterplot(data=iris,x='petal_length',y='petal_width',hue='species')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.title('Scatter Plot')
plt.show()
```

```python
# tips
plt.figure(figsize=(6,4))
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='smoker',palette=['#2B7FFF','#ED6AFF'])
plt.show()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='smoker',hue='smoker')
plt.show()
```

```python
# lineplots

# bivariate analysis
# relational graphs
# trends
# anomalies
# timeseries
```

```python
gap

ind = gap.query('country=="India"')

sns.lineplot(data=ind,x='year',y='pop')
plt.show()
```

```python
# india

neigbors = gap[gap.country.isin(['Nepal','China','India','Pakistan','Bangladesh','Afghanistan'])]

sns.lineplot(data=neigbors,x ='year',y='lifeExp',hue='country')
plt.title('Life Expectancy over the years in Neghbouring COuntries around India')
plt.show()
```

```python
# continents

cont = gap[gap.country.isin(['Canada','Argentina','Zimbabwe','Australia','China','Germany'])]

sns.lineplot(data=cont,x='year',y='gdpPercap',hue='country',style='continent')
plt.legend(loc='right')
```

```python
# page oriented
# figure level

sns.relplot(data=cont,x='year',y='gdpPercap',hue='country',style='continent',kind='line')
plt.show()
```

```python
palette = ['#3B82F6', '#F472B6']  # Blue, Pink
palette = ['#4C78A8', '#E78AC3']  # Muted Blue, Soft Pink
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',style='smoker',palette=palette,col='sex',row='smoker')
plt.show()
```

```python
# sns.histplot // axes level
# sns.kdeplot // axes level
# sns.rugplot // axes level
# sns.displot // figure level
# kind="kde", "rug"
```

```python
tips.total_bill
plt.figure(figsize=(12,4))
sns.histplot(data=tips,x='total_bill',bins=range(0,56,5),kde=True,hue='sex',element='step')
plt.title('Distribution of Total Bill')
plt.ylabel('frequency')
plt.show()
```

```python
tips.total_bill
plt.figure(figsize=(12,4))
sns.displot(data=tips,x='total_bill',bins=range(0,56,5),kde=True,hue='sex',element='step',col='sex')
plt.title('Distribution of Total Bill')
plt.ylabel('frequency')
plt.show()
```

```python
# kde --> kernel density estimator

plt.figure(figsize=(12,4))
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
plt.title('Distribution of Total Bill')
plt.show()
```
